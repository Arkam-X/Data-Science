```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 

warnings.filterwarnings("ignore")

df = pd.read_csv('IPL.csv')
df.head()
```

### Questions!!

#### 1. Team won the most matches.

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index, x = match_wins.values, palette='viridis')
plt.title('Most matches winned by Team')
plt.xlabel('Number of Matches')
plt.ylabel('Team Names')
```

#### 2. Toss Decision Trends.

```python
sns.countplot(x = df['toss_decision'], palette='rocket')
plt.title("Toss Decision")
plt.xlabel("Decision")
```

#### 3. Toss Winner vs Match Winner

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count / df.shape[0]) * 100
float(percentage.round(2))
```

#### 4. How did teams win? (Runs vs Wickets)

```python
sns.countplot(x = df['won_by'])
plt.title("Won By")
```

#### 5. Most matches played by Venue.

```python
venue_count = df['venue'].value_counts()
venue_count
sns.barplot(y = venue_count.index, x = venue_count.values, palette="rainbow")
```

### Key Player Performances.

#### 1. Most "Player of the Match" awards.

```python
count = df['player_of_the_match'].value_counts().head(10)
sns.barplot(x = count.values, y = count.index)
plt.title("Most Man of the Match awards")
plt.ylabel("Names")
plt.xlabel("Number of Awards")
```

#### 2. Top Scorers.

```python
df.head()
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(5)
high.plot(kind = 'barh')
```

#### 3. 10 Best Bowlers.

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
sns.barplot(y = top_bowlers.index, x = top_bowlers.values, palette='rocket')
plt.ylabel('Bowlers')
# top_bowlers.plot(kind='barh')
```
